```python
import numpy as np

heads = np.array([12.5, 12.1, 11.8, 11.2, 10.9])   # head (m) along a flow line
print(heads)
print(heads.shape)    # (5,)
print(heads.mean(), heads.max(), heads.min())
```

```python
# Maths on whole arrays (no loops needed)
distance = np.array([0, 100, 200, 300, 400])   # m

# gradient between neighbouring wells
dh = np.diff(heads)          # change in head
dx = np.diff(distance)       # change in distance
gradient = -dh / dx
print(gradient)
```

```python
# Reuse your Darcy function on arrays
def darcy_discharge(K, i, A):
    return K * i * A

Q = darcy_discharge(5.0, gradient, 100.0)
print(Q)
```

```python
# 2D arrays (a model grid)
grid = np.full((5, 5), 10.0)      # 5 rows x 5 columns, all heads = 10 m
grid[:, 0] = 15.0                  # left column = 15 m (fixed-head boundary)
print(grid)
print(grid.shape)
```

```python
# Indexing: grabbing single cells and blocks
print(grid[2, 3])
```

```python
print(grid[1:4, 1:4])
```

```python
# Filtering: picking values by a condition
K_field = np.array([0.1, 5.0, 12.0, 0.01, 30.0])
```

```python
print(K_field > 1.0)
```

```python
print(K_field[K_field > 1.0])
```

```python
print(np.where(K_field > 1.0, "high", "low"))
```

```python
import numpy as np

# 1. Grid with fixed-head boundaries
grid = np.full((10, 10), 20.0)
grid[:, 0] = 25.0      # left column
grid[:, -1] = 15.0     # right column (last column)
print(grid)
```

```python
# 2. Straight-line drop across every row
grid[:] = np.linspace(25, 15, 10)
print(grid)
```

```python
# 3. Head change between neighbouring columns in the first row
print(np.diff(grid[0, :]))
```

```python
# 4. Random conductivity values above 10
K = np.random.uniform(0.01, 50, 100)
print((K > 10).sum())
```
