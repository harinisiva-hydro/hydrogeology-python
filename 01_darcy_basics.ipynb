```python
K = 5.0      # hydraulic conductivity (m/day)
i = 0.02     # hydraulic gradient (dimensionless)
A = 100.0    # cross-sectional area (m^2)

Q = K * i * A
print(Q)
```

```python
# Defining function
def darcy_discharge(K, i, A):
    """Return discharge Q (m^3/day) from Darcy's Law: Q = K * i * A."""
    return K * i * A

print(darcy_discharge(5.0, 0.02, 100.0))
print(darcy_discharge(0.5, 0.01, 250.0))
```

```python
# Lists and Loops
materials = ["clay", "silt", "sand", "gravel"]
K_values = [0.001, 0.1, 10.0, 100.0]   # m/day, rough typical values

for name, K in zip(materials, K_values):
    Q = darcy_discharge(K, 0.01, 50.0)
    print(f"{name}: Q = {Q:.3f} m^3/day")
```

```python
# A decision with if
def classify_K(K):
    if K < 0.01:
        return "low permeability"
    elif K < 10:
        return "moderate permeability"
    else:
        return "high permeability"

for name, K in zip(materials, K_values):
    print(name, "->", classify_K(K))
```

```python
def seepage_velocity(K, i, n):
    """Return actual groundwater velocity (m/day): v = K * i / n."""
    if i <= 0:
        print(f"Warning: gradient i = {i} is zero or negative.")
    return K * i / n


materials = ["clay", "silt", "sand", "gravel"]
K_values = [0.001, 0.1, 10.0, 100.0]   # m/day
porosity = 0.4
gradient = 0.01

for name, K in zip(materials, K_values):
    v = seepage_velocity(K, gradient, porosity)
    print(f"{name}: v = {v:.6f} m/day")
```
